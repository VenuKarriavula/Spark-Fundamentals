# 01-lazy-evaluation-and-actions

Build a lazy plan, then execute count(). Expected result: 1 row for New York.

Run in Databricks with a provided spark session. Run cells from top to bottom. See ../docs/DEMO.md for setup and reset instructions.


```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

### **Lazy Evaluation & Action**


```python
data = [
    ("Alice", 25, "New York"),
    ("Bob", 30, "San Francisco"),
    ("Charlie", 35, "Chicago")
]

schema = StructType([
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True),
    StructField("city", StringType(), True)
])

df_new = spark.createDataFrame(data, schema=schema)
```

```python
df_new = df_new.filter(col('city')=='New York')
```

```python
df_new = df_new.select('city')
```

```python
df_new.count()
```
