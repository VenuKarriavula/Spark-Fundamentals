```python
from pyspark.sql.functions import * 
from pyspark.sql.types import *
spark.conf.set("spark.sql.adaptive.enabled",False)
```

```python
# Data Reading

dbutils.widgets.text("megamart_path", "", "Optional MegaMart CSV path")
megamart_path = dbutils.widgets.get("megamart_path").strip()
if megamart_path:
    df = spark.read.csv(megamart_path, header=True, inferSchema=True)
else:
    # Synthetic demo data; no external files or volume permissions required.
    df = spark.createDataFrame(
        [(1, "Sneakers"), (2, "Boots"), (3, "Sneakers"),
         (4, "Sandals"), (5, "Sneakers"), (6, "Boots")],
        "order_id INT, product_name STRING"
    )

# Narrow Transformations
df = df.filter(col("product_name")=='Sneakers')
df=df.select('order_id','product_name')

# Wide Transformations
df = df.groupBy('product_name').agg(count(col('order_id')))

display(df) # Action 3
```
