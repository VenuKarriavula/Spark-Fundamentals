# 02-transformations

Compare narrow transformations with repartition and grouped counts. Inspect Exchange operators; the final groups each contain 4 rows.

Run in Databricks with a provided spark session. Run cells from top to bottom. See ../docs/DEMO.md for setup and reset instructions.


```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python

from pyspark.sql import functions as F

# Disable AQE for predictable stage behavior
spark.conf.set("spark.sql.adaptive.enabled", "false")
spark.conf.set("spark.sql.shuffle.partitions", "4")

# Create 12 records with 4 input partitions
df = spark.range(0, 12, 1, 4)
```

##Narrow Transformations

```python
# Narrow transformation 1
df1 = df.filter("id > 2")
df2 = df1.select("id")

df1.show()
```

## Wide Transformations

```python
df3 = df2.repartition(2)

df3.collect()
```

```python
# Wide Transformation 1
df2 = df.repartition(2)

# Wide Transformation 2
df3 = df2.groupBy((F.col("id") % 3).alias("group_id")).count()

# Run only one action
display(df3)
```
