# 03-repartition-vs-coalesce

Compare eight input partitions reduced to two by repartition and coalesce. Inspect both physical plans.

Run in Databricks with a provided spark session. Run cells from top to bottom. See ../docs/DEMO.md for setup and reset instructions.


```python
df = spark.range(1000, numPartitions=8)
```

```python
df1 = df.repartition(2)

df1.explain()
df1.collect()
```

```python
df2 = df.coalesce(2)

df2.explain()
df2.collect()
```
