# 08-caching-execution

Cache and reuse the DataFrame. Inspect InMemoryTableScan / InMemoryRelation and clean up the cache at the end.

Run in Databricks with a provided spark session. Run cells from top to bottom. See ../docs/DEMO.md for setup and reset instructions.


```python
from pyspark.sql.functions import * 

# Create first DataFrame
data1 = [
    (1, "Alice"),
    (2, "Bob"),
    (3, "Charlie"),
    (4, "David"),
    (5, "Eva")
]
df1 = spark.createDataFrame(data1, ["id", "name"])
```

```python
display(df1)
```

```python
df1.cache()
```

```python
df2=df1.filter(col('id')==1)
```

```python
display(df2)
```

```python
df2.explain('extended')
```

```python
# Release demo cache after inspecting the plan.
df1.unpersist()
```
