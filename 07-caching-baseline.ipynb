```python
from pyspark.sql.functions import * 

# Create first DataFrame
data1 = [
    (1, "Alice"),
    (2, "Bob"),
    (3, "Charlie"),
    (4, "David"),
    (5, "Eva")
]
df1 = spark.createDataFrame(data1, ["id", "name"])
```

```python
display(df1)
```

```python
df1 = df1.withColumn('Flag',lit('Yes'))
```

```python
display(df1)
```

```python
df2=df1.filter(col('id')==1)
```

```python
display(df2)
```
